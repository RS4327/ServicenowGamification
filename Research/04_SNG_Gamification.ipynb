```python
class ConfigurationManager:
    def __init__(self,
                 CONFIG_PATH=CONFIG_FILE_PATH,
                 PARAMS_PATH=PARAMS_FILE_PATH
                 ):
         print(CONFIG_PATH)
                 
         print(f"path config yaml {Path(CONFIG_PATH)}")
         self.config=Read_Yaml(Path(CONFIG_PATH))
         self.params=Read_Yaml(Path(PARAMS_PATH))
         self.Data_Gamification = self.config.Data_Gamification
    def get_datagamificationconfig(self)->DataGamificationConfig:
         Config=self.Data_Gamification
         create_directories([Config.root_dir])
         data_gamification_config=DataGamificationConfig(
              root_dir =Config.root_dir,
              data_path=Config.data_path,
              local_data_file=Config.local_data_file
         )
         return data_gamification_config


```

```python
class DataGamification:

    def __init__(self, config: DataGamificationConfig):
        self.config = config
        self.params = Read_Yaml(PARAMS_FILE_PATH)

    def Gamification(self):

        # Read preprocessed dataset
        df = pd.read_csv(self.config.data_path)

        def calculate_sla_status(row):

            priority = str(row["priority"])
            #print(f"priority {priority}")

            if priority == "1 - Critical":
                return "Met" if row["resolution_hours"] <= self.params.sla.priority_1 else "Breached"

            elif priority == "2 - High":
                return "Met" if row["resolution_hours"] <= self.params.sla.priority_2 else "Breached"

            elif priority == "3 - Moderate":
                return "Met" if row["resolution_hours"] <= self.params.sla.priority_3 else "Breached"

            elif priority == "4 - Low":
                return "Met" if row["resolution_hours"] <= self.params.sla.priority_4 else "Breached"

            else:
                return "5 - Planning"

        df["sla_status"] = df.apply(calculate_sla_status, axis=1)
        local_path=self.config.local_data_file
        print(f"Preprocessed data saved to: {local_path}")
        df.to_csv(local_path, index=False)

        sla_compliance = (df["sla_status"] == "Met").mean() * 100

        print(f"SLA Compliance: {sla_compliance:.2f}%")
        print(df[["number", "sla_status","resolution_hours"]].head())
```

```python
config = ConfigurationManager()

gamification_config = config.get_datagamificationconfig()

gamification = DataGamification(gamification_config)   
df=gamification.Gamification()
```
