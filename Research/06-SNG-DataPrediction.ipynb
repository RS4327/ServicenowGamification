```python
class ConfigurationManager:
    def __init__(self,
                 CONFIG_PATH=CONFIG_FILE_PATH,
                 PARAMS_PATH=PARAMS_FILE_PATH
                 ):
         print(CONFIG_PATH)
                 
         print(f"path config yaml {Path(CONFIG_PATH)}")
         self.config=Read_Yaml(Path(CONFIG_PATH))
         self.params=Read_Yaml(Path(PARAMS_PATH))
         self.Data_Prediction = self.config.Data_Prediction
     
    def get_dataprediction(self)->DataPredictionConfig:
         Config=self.Data_Prediction
         create_directories([Config.root_dir])
         data_prediction_config=DataPredictionConfig(
              root_dir=Config.root_dir,
              data_path=Config.data_path,
              model_path=Config.model_path, 
              scaler_path =Config.scaler_path, 
              prediction_output =Config.prediction_output, 
              probability_threshold =Config.probability_threshold 
         )
         return data_prediction_config
```

```python
import os 
from pathlib import Path
from ServicenowGamification.Utils.Common import *

from ServicenowGamification.Constant import CONFIG_FILE_PATH, PARAMS_FILE_PATH
from ServicenowGamification.Entity.Config_Entity import  DataIngestionConfig 
from ServicenowGamification.Entity.Config_Entity import DataPreProcessingConfig
from ServicenowGamification.Entity.Config_Entity import DataMetricsCalculationConfig
from ServicenowGamification.Entity.Config_Entity import DataGamificationConfig
from ServicenowGamification.Entity.Config_Entity import DataMachineLearningModelsConfig
from ServicenowGamification.Entity.Config_Entity import DataPredictionConfig


class ConfigurationManager:
    def __init__(self,
                 CONFIG_PATH=CONFIG_FILE_PATH,
                 PARAMS_PATH=PARAMS_FILE_PATH

                 ):
         
        #  ROOT_DIR = Path(__file__).resolve().parents[3]
        #          # Read configuration files
        #  #print("ROOT_DIR :", ROOT_DIR)
         
        #  config_path = ROOT_DIR / "Config" / "Config.yaml"
        #  params_path = ROOT_DIR / "Params.yaml"
         
        #  #print("Config Path :", config_path)
        #  #print("Config Exists :", config_path.exists())
         
         print(CONFIG_PATH)
                          
         print(f"path config yaml {Path(CONFIG_PATH)}")
         self.config=Read_Yaml(Path(CONFIG_PATH))
         self.params=Read_Yaml(Path(PARAMS_PATH))
                 
                   
         
         #print("Current Working Directory:", os.getcwd())
         
         for p in Path(".").rglob("*.yaml"):
            print("Found YAML:", p.resolve())
         create_directories([self.config.artifacts_root]) 
        # print(f"Parent Directory {self.config.artifacts_root}")

         self.Data_Ingestion = self.config.Data_Ingestion

         self.Data_PreProcessing = self.config.Data_PreProcessing
         self.Data_MetricsCalculation = self.config.Data_MetricsCalculation
         self.Data_Gamification=self.config.Data_Gamification
         self.Data_MachineLearningModels = self.config.Data_MachineLearningModels
         self.Data_Prediction = self.config.Data_Prediction
        #  print("Artifacts Root:", self.config.artifacts_root)
        #  print("Data Ingestion Root:", self.Data_Ingestion.root_dir)
        #  print("Data PreProcessing Root:", self.Data_PreProcessing.root_dir)
        #  print("Data Metrics Root:", self.Data_MetricsCalculation.root_dir)

        
    def get_data_ingestion_config(self)->DataIngestionConfig:
        Config=self.Data_Ingestion
        create_directories([Config.root_dir])
        data_ingestion_config=DataIngestionConfig(
            root_dir=Config.root_dir,
            source_url=Config.source_url,
            user_name=Config.user_name,
            password=Config.password,
            local_data_file=Config.local_data_file

        )

        return data_ingestion_config
    def get_data_preprocessing_config(self)->DataPreProcessingConfig:
             
        Config=self.Data_PreProcessing
        create_directories([Config.root_dir])
        data_preprocessingconfig=DataPreProcessingConfig(
            root_dir=Config.root_dir,
            data_path=Config.data_path,
            local_data_file=Config.local_data_file
        )
        return data_preprocessingconfig
    def get_metricscalculationconfig(self)->DataMetricsCalculationConfig:
        Config=self.Data_MetricsCalculation
        create_directories([Config.root_dir])
        data_preprocessingconfig=DataMetricsCalculationConfig(
            root_dir=Config.root_dir,
             data_path=Config.data_path,
            local_data_file=Config.local_data_file
        )
        return data_preprocessingconfig
    def get_datagamificationconfig(self)->DataGamificationConfig:
             Config=self.Data_Gamification
             create_directories([Config.root_dir])
             data_gamification_config=DataGamificationConfig(
                  root_dir =Config.root_dir,
                  data_path=Config.data_path,
                  local_data_file=Config.local_data_file,
                  incident_points=Config.incident_points,
                  analyst_points=Config.analyst_points,
                  badgeresults=Config.badgeresults,
                  badgesummary=Config.badgesummary,
                  levelresults=Config.levelresults,
                  levelsummary= Config.levelsummary,
                  leaderboardfile= Config.leaderboardfile, 
                  topfile= Config.topfile, 
                  bottomfile= Config.bottomfile,
                  teamfile= Config.teamfile,
                  groupfile= Config.groupfile ,
                  achievementfile= Config.achievementfile ,
                  summaryfile= Config.summaryfile   ,
                  monthlyranking= Config.monthlyranking ,
                  monthlytopfile= Config.monthlytopfile ,
                  monthlysummaryfile= Config.monthlysummaryfile            
             )
             return data_gamification_config
    
    def get_data_machinelearningmodels(self)->DataMachineLearningModelsConfig:
            Config=self.Data_MachineLearningModels
            create_directories([Config.root_dir])
            data_machinelearningmodels_config=DataMachineLearningModelsConfig(
                root_dir=Config.root_dir,
                data_path=Config.data_path, 
                X_train =Config.X_train, 
                X_test =Config.X_test, 
                y_train =Config.y_train, 
                y_test =Config.y_test,
                scalerpkl =Config.scalerpkl, 
                Xtrainengineered =Config.Xtrainengineered, 
                Xtestengineered =Config.Xtestengineered  
            )
            return data_machinelearningmodels_config
    
    def get_dataprediction(self)->DataPredictionConfig:
             Config=self.Data_Prediction
             create_directories([Config.root_dir])
             data_prediction_config=DataPredictionConfig(
                  root_dir=Config.root_dir,
                  data_path=Config.data_path,
                  model_path=Config.model_path, 
                  scaler_path =Config.scaler_path, 
                  prediction_output =Config.prediction_output, 
                  probability_threshold =Config.probability_threshold 
             )
             return data_prediction_config


```

```python

class PredictSLABreach:
    """
    Predict SLA Breaches using trained XGBoost model.
    """

    def __init__(self, config:DataPredictionConfig):
        self.config = config
        config_manager=ConfigurationManager()
        self.model = None
        self.scaler = None
        self.params = Read_Yaml(PARAMS_FILE_PATH)
        # self.preprocessing =DataPreProcessing()
        # self.MetricsCal=MetricsCalculation()
        # self.level=Gamification()


            # Pass the preprocessing configuration
       
        self.preprocessing = DataPreProcessing(config_manager.get_data_preprocessing_config() )

        self.metrics = MetricsCalculation(config_manager.get_metricscalculationconfig())

        self.gamification = Gamification(config_manager.get_datagamificationconfig() )
        self.encoding = MachineLearningModels(config_manager.get_data_machinelearningmodels() )

    ###############################################################
    # Load Model
    ###############################################################

    def load_model(self):
        print('###########################################################')
        print('# Load Model')
        print('###########################################################')
        model_path=self.config.model_path
        print( 'model path ' ,model_path)
        print(os.path.exists(self.config.model_path))
        scaler_path=self.config.scaler_path
        self.model = joblib.load(model_path)
        self.scaler = joblib.load(scaler_path)
        print("Model Loaded Successfully.")

    def datacleaning(self):

        df = pd.read_csv(self.config.data_path)
        print(df.columns)
        df = self.preprocessing.Datapreprocessing(df)

        return df 

    def mttr(self,df):
        df=self.metrics.MTTR(df)
        df,art=self.metrics.calculate_art(df)
        df=self.metrics.calculate_sla(df)
        df=self.metrics.calculate_fcr(df)
        df,metrics=self.metrics.performance_metrics(df)


        return df
         
    def levl(self, df):

        print(type(df))

        df = self.gamification.calculate_points(df)
        print("calculate_points :", type(df))

        df_f, df_analyst_summary = self.gamification.p_save(df)
        print("p_save :", type(df_f), type(df_analyst_summary))

        df_b = self.gamification.assign_badges(df_f)
        print("assign_badges :", type(df_b))

        df = self.gamification.b_save(df_b)
        print("b_save :", type(df))

        df_l = self.gamification.calculate_level(df)
        print("calculate_level :", type(df_l))

        df = self.gamification.l_save(df_l)
        print("l_save :", type(df))

        df_lead = self.gamification.generate_leaderboard(df_analyst_summary)
        print("generate_leaderboard :", type(df_lead))

        df = self.gamification.lead_save(df_lead)
        print("lead_save :", type(df))

        df_ach = self.gamification.calculate_achievements(df_lead)
        print("calculate_achievements :", type(df_ach))

        df = self.gamification.ach_save(df_ach)
        print("ach_save :", type(df))

        df_m = self.gamification.generate_monthly_ranking(df_f)
        print("generate_monthly_ranking :", type(df_m))

        df = self.gamification.mon_save(df_m)
        print("mon_save :", type(df))

        return df_l
    def encodings(self,df):
        df_clean=self.encoding.clean_data(df)
        df_fea=self.encoding.select_features(df_clean)
        df_enc=self.encoding.encode_features(df_fea)
        return df_enc
        # X_train,X_test,y_train, y_test=data_mlm.split_dataset(df_enc)
        # data_mlm.train_test_save(X_train,X_test,y_train, y_test)
        # train_df,test_df=data_mlm.create_time_features()
        # train_df,test_df=data_mlm.create_interaction_features(train_df,test_df)
        # train_df,test_df=data_mlm.log_transform(train_df,test_df)
        # train_df,test_df,scaler=data_mlm.scale_features(train_df,test_df)
        # data_mlm.save_scaler(scaler)
        # data_mlm.save_dataset(train_df,test_df)
        # models=data_mlm.get_models()
        # ml_model_name,best_accuracy,best_model,results=data_mlm.ModelsTraining(models)

   
    def preprocess(self,df):

        feature_columns = [             
            "priority",
            "assignment_group",
            "assigned_to",
            "resolution_hours",
            "reopen_count",
            "reassignment_count",
            "points",
            "level"
        ]

        df = df[feature_columns]
        df["resolution_reopen"] = (df["resolution_hours"] * (df["reopen_count"] + 1))




        numeric_columns = df.select_dtypes( include=["int64", "float64"]).columns
        print('numeric_columns ',numeric_columns)
        df[numeric_columns] = self.scaler.transform(df[numeric_columns] )
        return df

    ###############################################################
    # Prediction
    ###############################################################

    def predict(self,df):
        print('###########################################################')
        print('# Prediction')
        print('###########################################################')    
        probability = self.model.predict_proba(df )[:, 1]
        prediction = self.model.predict(df )

        df["breach_probability"] = probability
        df["prediction"] = prediction
        df["prediction"] = df["prediction"].map(
            {
                0: "SLA Met",
                1: "SLA Breach"
            }
        )
        return df

    ###############################################################
    # Risk Category
    ###############################################################

    def risk_level(self,df):
        print('###########################################################')
        print('# Risk Category')
        print('###########################################################')           

        def category(probability):
            if probability >= 0.80:
                return "High"
            elif probability >= 0.50:
                return "Medium"
            else:
                return "Low"
        df["risk_level"] = df["breach_probability"].apply(category)
        return df

    ###############################################################
    # Recommendation
    ###############################################################

    def recommendation(self,df):
        print('###########################################################')
        print('# recommendation')
        print('###########################################################')           

        recommendation = []
        for _, row in df.iterrows():

            if row["risk_level"] == "High":
                recommendation.append( "Escalate Immediately" )

            elif row["risk_level"] == "Medium":
                recommendation.append("Monitor Closely")
            else:
                recommendation.append( "No Action Required" )

        df["recommendation"] = recommendation

        return df

    ###############################################################
    # Save Prediction
    ###############################################################

    def save_prediction(self,df):
        output_folder = self.config.prediction_output
      
        path = os.path.join( output_folder, "SLAPredictions.csv")

        df.to_csv( path, index=False)

        print(f"Prediction Saved : {path}")
        return df


```

```python

config = ConfigurationManager()
data_predict_mod = config.get_dataprediction()
PredictSLAB=PredictSLABreach(data_predict_mod)
df_load=PredictSLAB.load_model()
df=PredictSLAB.datacleaning()
df=PredictSLAB.mttr(df)
df=PredictSLAB.levl(df)
df=PredictSLAB.encodings(df)
df=PredictSLAB.preprocess(df)
df=PredictSLAB.predict(df)
df=PredictSLAB.risk_level(df)
df=PredictSLAB.recommendation(df)
df=PredictSLAB.save_prediction(df)
 




```
